```python
#largest in the list
no=input("enter  the list of numbers:").split(",")
largest=no[0]
for num in no:
    if num>largest:
        largest=num
print("largest number is:",largest)
```

```python
#remove duplicate
num=input("enter the list of numbers:").split()
og=[]
for i in num:
    if i not in og:
        og.append(i)

print("non-duplicated list is:",og)
```

```python

#common members
def common_member(list1, list2):
    for item1 in list1:
        for item2 in list2:
            if item1 == item2:
                return True
    return False


list1 = [1, 2, 3, 4]
list2 = [5, 6, 3, 8]

print(common_member(list1, list2))
```
